# Sistema de Recomendação com Aprendizado por Reforço — v2

Este notebook implementa um sistema de recomendação cooperativo com **cinco agentes Actor-Critic paralelos**, treinados usando `gymnasium.Env` e PyTorch puro.

## Melhorias da v2 em relação à v1

| Componente | v1 | v2 |
|-----------|----|----|
| Observação | média de embeddings | + distribuições de diversidade do histórico |
| Seleção de filme | dot-product soft | dois estágios: amostra alvo → boost 3× |
| Recompensa | calculada em `genre_p` (prob.) | calculada no **filme selecionado** |
| Cold start | retorna 0.5 fixo | usa priors reais do catálogo |
| Entropia | `coef=0.01` | `coef=0.05` — mais exploração |
| Temperatura | 1.5 → 1.0 | **2.0 → 1.0** |
| Normalização | nenhuma | Welford online por agente |

## Arquitetura

```
        ┌──────────────────────────── Estado Compartilhado ────────────────────────────┐
        │  mean(features 50 filmes do usuário) + mean(features 500 recomendados)       │
        │  + distribuições atuais de gênero/raça/região/idioma no histórico            │
        │  + rating médio do usuário por gênero                                        │
        └───────────┬──────────────┬──────────────┬──────────────┬─────────────────────┘
                    ▼              ▼              ▼              ▼              ▼
               Agente 1       Agente 2       Agente 3       Agente 4       Agente 5
               Gênero      Gênero Dir.     Raça Dir.     Região Dir.      Idioma
              (n_genres)    [M|F|MIX]    [W|NW|MIX]    (9 regiões)     (n_langs)
```

## 4. Mapeamento de Grupos de Diversidade

Os Agentes 2, 3 e 4 trabalham com **grupos agregados** derivados das colunas de encoding:

| Agente | Grupos | Colunas agregadas |
|--------|--------|-----------------|
| 2 — Gênero do Dir. | MALE, FEMALE, MIX | `dir_gender_*` agrupadas por padrão no nome |
| 3 — Raça do Dir. | WHITE, NO_WHITE, MIX | `dir_race_*` agrupadas por padrão no nome |
| 4 — Região do Dir. | 9 regiões | `dir_region_*` diretas (1 coluna = 1 região) |

O agrupamento é feito por **correspondência de padrão no nome da coluna**, tornando o código robusto a variações de nomenclatura.

````python
# ── Agente 2: grupos de gênero do diretor ─────────────────────────────────────
GENDER_GROUPS = ['MALE', 'FEMALE', 'MIX']

def _match_gender(col: str) -> Optional[int]:
    c = col.replace('dir_gender_', '').lower()
    if '_ef' in c or '_mf' in c:   return 1   # FEMALE (antes de male!)
    if '_em' in c or '_mm' in c:   return 0   # MALE
    if '_un' in c or '_eg' in c:   return 2   # MIX
    return None

GENDER_COL_GROUP = {c: _match_gender(c) for c in gender_cols}
print('Mapeamento de gênero do diretor:')
for col, grp in GENDER_COL_GROUP.items():
    label = GENDER_GROUPS[grp] if grp is not None else '?'
    print(f'  {col:<45} → {label}')

# ── Agente 3: grupos de raça do diretor ──────────────────────────────────────
RACE_GROUPS = ['WHITE', 'NO_WHITE', 'MIX']

def _match_race(col: str) -> Optional[int]:
    c = col.replace('dir_race_', '').lower()
    if 'no_white' in c or 'nonwhite' in c:   return 1   # NO_WHITE (antes de white!)
    if 'white' in c:                          return 0   # WHITE
    if 'unknown' in c or 'mixed' in c:        return 2   # MIX
    return None

RACE_COL_GROUP = {c: _match_race(c) for c in race_cols}
print('\nMapeamento de raça do diretor:')
for col, grp in RACE_COL_GROUP.items():
    label = RACE_GROUPS[grp] if grp is not None else '?'
    print(f'  {col:<45} → {label}')

# ── Agente 4: região do diretor (direto das colunas) ─────────────────────────
REGION_NAMES = [c.replace('dir_region_', '').replace('_', ' ').title() for c in region_cols]
print(f'\nRegiões ({len(region_cols)}): {REGION_NAMES}')
````

## 5. Embeddings

Variáveis textuais e categóricas são transformadas em **vetores densos** antes do treinamento:

| Feature | Tipo | Dimensão | Técnica |
|---------|------|----------|---------|
| Título | texto | 32 | `nn.EmbeddingBag` sobre tokens de palavras |
| Gênero do filme | multi-hot | 16 | Projeção linear treinável |
| Idioma do filme | multi-hot | 8 | Projeção linear treinável |
| Dir. gênero/raça/região | multi-hot | direto | Sem projeção (já numérico) |

As projeções são inicializadas aleatoriamente e pré-computadas uma vez. Em produção,
poderiam ser ajustadas conjuntamente com o treinamento dos agentes.

````python
# ── Vocabulário para título ────────────────────────────────────────────────────
def _tokenize(text: str) -> list[str]:
    return re.sub(r'[^a-z0-9 ]', '', str(text).lower()).split()

word_counter = Counter()
for title in enc_df['title'].fillna(''):
    word_counter.update(_tokenize(title))

VOCAB = ['<PAD>', '<UNK>'] + [w for w, _ in word_counter.most_common(8000)]
WORD2IDX = {w: i for i, w in enumerate(VOCAB)}
print(f'Vocabulário: {len(VOCAB)} tokens')

# ── Dimensões dos embeddings ──────────────────────────────────────────────────
TITLE_DIM  = 32
GENRE_DIM  = 16
LANG_DIM   = 8

N_GENRES  = len(genre_cols)
N_LANGS   = len(lang_cols)
N_GENDER  = len(gender_cols)
N_RACE    = len(race_cols)
N_REGION  = len(region_cols)

# Dimensão do embedding denso por filme
MOVIE_FEAT_DIM = TITLE_DIM + GENRE_DIM + LANG_DIM + N_GENDER + N_RACE + N_REGION

# Dimensão das estatísticas de diversidade adicionadas à observação:
#   distribuição de gênero + raça + região + idioma no sys_hist + preferência do usuário por gênero
STATS_DIM = N_GENRES + 3 + 3 + N_REGION + N_LANGS + N_GENRES

# Dimensão total da observação
OBS_DIM = 2 * MOVIE_FEAT_DIM + STATS_DIM
print(f'Feature dim por filme : {MOVIE_FEAT_DIM}')
print(f'Stats dim             : {STATS_DIM}')
print(f'Observation dim total : {OBS_DIM}')


class MovieEmbedder(nn.Module):
    """
    Converte features brutas de um filme em um vetor denso (MOVIE_FEAT_DIM).

    Componentes:
      - title_emb   : EmbeddingBag sobre tokens de palavras do título
      - genre_proj  : projeção linear do multi-hot de gêneros
      - lang_proj   : projeção linear do multi-hot de idiomas
      - dir_*       : vetores de encoding do diretor usados diretamente
    """
    def __init__(self):
        super().__init__()
        self.title_emb  = nn.EmbeddingBag(len(VOCAB), TITLE_DIM, mode='mean', padding_idx=0)
        self.genre_proj = nn.Linear(N_GENRES, GENRE_DIM)
        self.lang_proj  = nn.Linear(N_LANGS,  LANG_DIM)

    def forward(self, title_tokens, genre_vec, lang_vec, dir_vec):
        """title_tokens: (B, T) | genre_vec: (B, N_GENRES) | ..."""
        t_emb = self.title_emb(title_tokens)          # (B, TITLE_DIM)
        g_emb = F.relu(self.genre_proj(genre_vec))    # (B, GENRE_DIM)
        l_emb = F.relu(self.lang_proj(lang_vec))      # (B, LANG_DIM)
        return torch.cat([t_emb, g_emb, l_emb, dir_vec], dim=-1)  # (B, MOVIE_FEAT_DIM)


embedder = MovieEmbedder()
print(f'\nMovieEmbedder criado. Parâmetros: {sum(p.numel() for p in embedder.parameters()):,}')
````

````python
def _gender_group_vec(gender_row: np.ndarray) -> np.ndarray:
    """Agrega colunas de gênero do diretor em 3 grupos: [MALE, FEMALE, MIX]."""
    vec = np.zeros(3, dtype=np.float32)
    for i, col in enumerate(gender_cols):
        grp = GENDER_COL_GROUP.get(col)
        if grp is not None:
            vec[grp] += float(gender_row[i])
    return vec

def _race_group_vec(race_row: np.ndarray) -> np.ndarray:
    """Agrega colunas de raça do diretor em 3 grupos: [WHITE, NO_WHITE, MIX]."""
    vec = np.zeros(3, dtype=np.float32)
    for i, col in enumerate(race_cols):
        grp = RACE_COL_GROUP.get(col)
        if grp is not None:
            vec[grp] += float(race_row[i])
    return vec


def build_movie_features(df: pd.DataFrame) -> dict:
    """
    Pré-computa vetores de features para todos os filmes.
    Retorna {movieid: dict_de_features}.
    """
    embedder.eval()
    features = {}

    for _, row in df.iterrows():
        try:
            mid = int(row['movieid'])
        except (ValueError, TypeError):
            continue

        words  = _tokenize(str(row.get('title', '')))
        tokens = [WORD2IDX.get(w, 1) for w in words] or [0]

        genre_v  = row[genre_cols].values.astype(np.float32)
        lang_v   = row[lang_cols].values.astype(np.float32)
        gender_v = row[gender_cols].values.astype(np.float32)
        race_v   = row[race_cols].values.astype(np.float32)
        region_v = row[region_cols].values.astype(np.float32)
        dir_v    = np.concatenate([gender_v, race_v, region_v])

        with torch.no_grad():
            feat_vec = embedder(
                torch.tensor([tokens], dtype=torch.long),
                torch.tensor([genre_v]),
                torch.tensor([lang_v]),
                torch.tensor([dir_v]),
            ).squeeze(0).numpy()

        features[mid] = {
            'vec':          feat_vec,
            'genres':       genre_v,
            'langs':        lang_v,
            'gender_raw':   gender_v,
            'race_raw':     race_v,
            'region':       region_v,
            'gender_group': _gender_group_vec(gender_v),
            'race_group':   _race_group_vec(race_v),
        }

    return features


print('Pré-computando features dos filmes (pode levar ~1-2 min)...')
movie_features = build_movie_features(enc_df)
all_movie_ids  = list(movie_features.keys())
print(f'Pronto! {len(movie_features):,} filmes com vetores de {MOVIE_FEAT_DIM} dimensões.')


# ── Priors do dataset ─────────────────────────────────────────────────────────
# Distribuição real de cada atributo em todo o catálogo.
# Usados como fallback quando sys_hist está vazio (cold start), substituindo o
# valor fixo 0.5 que não fornece gradiente de aprendizado.

def _mean_norm(vecs_list: list, n: int) -> np.ndarray:
    if not vecs_list:
        return np.ones(n, dtype=np.float32) / n
    m = np.mean(vecs_list, axis=0)
    s = m.sum()
    return (m / s if s > 0 else np.ones(n) / n).astype(np.float32)

DATASET_PRIORS = {
    'genre':  _mean_norm([f['genres']       for f in movie_features.values() if f['genres'].sum() > 0], N_GENRES),
    'gender': _mean_norm([f['gender_group'] for f in movie_features.values() if f['gender_group'].sum() > 0], 3),
    'race':   _mean_norm([f['race_group']   for f in movie_features.values() if f['race_group'].sum() > 0], 3),
    'region': _mean_norm([f['region']       for f in movie_features.values() if f['region'].sum() > 0], N_REGION),
    'lang':   _mean_norm([f['langs']        for f in movie_features.values() if f['langs'].sum() > 0], N_LANGS),
}

print('\nPriors do dataset (distribuição real do catálogo):')
print(f"  Gênero dir.: {dict(zip(GENDER_GROUPS, DATASET_PRIORS['gender'].round(3)))}")
print(f"  Raça dir.  : {dict(zip(RACE_GROUPS,   DATASET_PRIORS['race'].round(3)))}")
````

## 6. Carregador de Dados

O `DataLoader` organiza as estruturas de acesso rápido:
- `ratings` — dicionário `{user_id → {movie_id → rating}}`
- `rec_history` — dicionário `{user_id → [movie_ids recomendados]}`
- Split 80/20 entre usuários de treino e teste

````python
def build_ratings(df: pd.DataFrame) -> dict:
    """Constrói {user_id: {movie_id: rating}} a partir do DataFrame."""
    out: dict[int, dict[int, float]] = defaultdict(dict)
    for _, row in df.iterrows():
        try:
            uid = int(row['USERID'])
            mid = int(row['MOVIEID'])
            rat = float(row['RATING'])
        except (ValueError, TypeError, KeyError):
            continue
        out[uid][mid] = rat
    return dict(out)


def build_rec_history(df: Optional[pd.DataFrame]) -> dict:
    """Constrói {user_id: [movie_ids recomendados]} a partir do DataFrame de recomendações."""
    if df is None:
        return {}
    out: dict[int, list[int]] = defaultdict(list)
    mid_col = next((c for c in df.columns if 'MOVIE' in c or 'ITEM' in c), None)
    if mid_col is None:
        return {}
    for _, row in df.iterrows():
        try:
            uid = int(row['USERID'])
            mid = int(row[mid_col])
            out[uid].append(mid)
        except (ValueError, TypeError):
            continue
    return dict(out)


ratings     = build_ratings(ratings_df)
rec_history = build_rec_history(recs_df)
all_users   = list(ratings.keys())

# Split 80/20
random.shuffle(all_users)
split = int(len(all_users) * 0.8)
train_users = all_users[:split]
test_users  = all_users[split:]

print(f'Usuários totais : {len(all_users):,}')
print(f'  Treino        : {len(train_users):,}')
print(f'  Teste         : {len(test_users):,}')
````

## 7. Ambiente `gymnasium.Env`

### Estado (observação estendida)

```
obs = [ mean(features dos 50 últimos filmes do usuário)    ]  ← MOVIE_FEAT_DIM
    + [ mean(features dos 500 últimos filmes recomendados) ]  ← MOVIE_FEAT_DIM
    + [ distribuição de gêneros de filmes no sys_hist      ]  ← N_GENRES
    + [ distribuição de gênero do diretor no sys_hist      ]  ← 3
    + [ distribuição de raça do diretor no sys_hist        ]  ← 3
    + [ distribuição de região do diretor no sys_hist      ]  ← N_REGION
    + [ distribuição de idiomas no sys_hist                ]  ← N_LANGS
    + [ rating médio do usuário por gênero (de ratings)    ]  ← N_GENRES
```

As estatísticas de diversidade tornam o que está **sobre-representado no histórico diretamente visível** para os agentes — sem precisar aprender a inferir isso indiretamente.

### Ação

```python
actions = {
    'genre':    array(n_genres,)   # Agente 1
    'gender':   array(3,)          # Agente 2 — [MALE, FEMALE, MIX]
    'race':     array(3,)          # Agente 3 — [WHITE, NO_WHITE, MIX]
    'region':   array(n_regions,)  # Agente 4
    'language': array(n_langs,)    # Agente 5
}
```

### Seleção de filme (dois estágios)

1. **Amostragem de alvos**: cada agente de diversidade amostra uma categoria a partir de sua distribuição (ex: Agente 2 sorteia MALE, FEMALE ou MIX com `np.random.choice(p=gender_p)`).
2. **Pontuação com boost**: filmes que pertencem às categorias alvo recebem pontuação 3× maior.

Isso cria **acoplamento direto** entre ação e resultado — o agente que diz "quero FEMALE" realmente obtém filmes com diretoras.

### Recompensas (v2)

> **Mudança fundamental**: as recompensas são calculadas sobre o **filme efetivamente selecionado**, não sobre a distribuição de probabilidade emitida. A versão anterior recompensava `gender_p = [0.1, 0.8, 0.1]` independentemente de qual filme foi escolhido — um desacoplamento que impedia o aprendizado de diversidade.

| Agente | Recompensa |
|--------|-----------|
| R1 — Gênero (filme) | `0.6 × pref_usuário + 0.4 × raridade_do_gênero_na_sessão` |
| R2 — Gênero Dir. | `0.6 × sub_representação_no_histórico + 0.4 × alinhamento_com_alvo` |
| R3 — Raça Dir. | idem |
| R4 — Região Dir. | idem |
| R5 — Idioma | `0.6 × sub_representação_no_histórico + 0.4 × compatibilidade_com_lang_p` |

**sub-representação** = `1 − dot(distribuição_do_filme, distribuição_histórica)` — valor alto quando o grupo do filme é raro no histórico.</cell>

````python
def _softmax(x: np.ndarray) -> np.ndarray:
    e = np.exp(x - x.max())
    return e / e.sum()


# Fator de decaimento por posição de recência (ajustável)
HIST_DECAY = 0.95   # posição 0 = peso 1.0, posição 1 = 0.95, posição 2 = 0.90, ...


class MovieRecommendationEnv(gym.Env):
    """
    Ambiente de recomendação cooperativo com 5 agentes paralelos — v2.

    Principais mudanças em relação à v1:
    - Observação estendida inclui distribuições de diversidade do sys_hist.
    - Histórico representado por média ponderada por recência (HIST_DECAY).
    - Seleção em dois estágios: amostra alvos → boost 3× para filmes correspondentes.
    - Recompensas calculadas sobre o FILME SELECIONADO (não sobre probabilidades).
    - Cold start usa priors reais do catálogo (não retorna 0.5 fixo).
    """

    HIST_USER = 50
    HIST_SYS  = 500
    MAX_STEPS = 10
    MAX_CAND  = 1000

    def __init__(self, features, ratings, rec_history, user_ids):
        super().__init__()
        self.features    = features
        self.ratings     = ratings
        self.rec_history = rec_history
        self.user_ids    = user_ids

        self.observation_space = spaces.Box(
            -np.inf, np.inf, shape=(OBS_DIM,), dtype=np.float32
        )
        self.action_space = spaces.Dict({
            'genre':    spaces.Box(0., 1., shape=(N_GENRES,),  dtype=np.float32),
            'gender':   spaces.Box(0., 1., shape=(3,),          dtype=np.float32),
            'race':     spaces.Box(0., 1., shape=(3,),          dtype=np.float32),
            'region':   spaces.Box(0., 1., shape=(N_REGION,),   dtype=np.float32),
            'language': spaces.Box(0., 1., shape=(N_LANGS,),    dtype=np.float32),
        })

        self._user_id          = None
        self._user_hist        = deque(maxlen=self.HIST_USER)
        self._sys_hist         = deque(maxlen=self.HIST_SYS)
        self._seen             = set()
        self._step_n           = 0
        self._last_targets     = {}
        self._genre_avg_rating = np.full(N_GENRES, 0.5, dtype=np.float32)

    # ── Observação ────────────────────────────────────────────────────────────

    def _weighted_mean_feat(self, movie_ids, most_recent_first: bool) -> np.ndarray:
        """
        Média ponderada por recência dos vetores de features.

        Filmes mais recentes recebem peso HIST_DECAY^0 = 1.0;
        o seguinte HIST_DECAY^1 = 0.95; e assim por diante.

        most_recent_first=True  → _sys_hist  (appendleft: índice 0 é o mais novo)
        most_recent_first=False → _user_hist (filled oldest-first: índice -1 é o mais novo)
        """
        ids = list(movie_ids)
        if not most_recent_first:
            ids = ids[::-1]   # inverte para que índice 0 = mais recente

        vecs    = []
        weights = []
        for i, mid in enumerate(ids):
            if mid in self.features:
                vecs.append(self.features[mid]['vec'])
                weights.append(HIST_DECAY ** i)

        if not vecs:
            return np.zeros(MOVIE_FEAT_DIM, dtype=np.float32)

        w = np.array(weights, dtype=np.float32)
        w /= w.sum()
        return (w[:, None] * np.stack(vecs)).sum(axis=0).astype(np.float32)

    def _get_dist(self, feat_key: str, n: int, prior_key: str) -> np.ndarray:
        """
        Distribuição do atributo nos últimos 200 filmes do sys_hist.
        Retorna DATASET_PRIORS[prior_key] se o histórico estiver vazio.
        """
        sys_list = list(self._sys_hist)[:200]
        if not sys_list:
            return DATASET_PRIORS[prior_key].copy()
        vecs = [self.features[m][feat_key] for m in sys_list if m in self.features]
        if not vecs:
            return DATASET_PRIORS[prior_key].copy()
        m = np.mean(vecs, axis=0)
        s = m.sum()
        return (m / s if s > 0 else DATASET_PRIORS[prior_key]).astype(np.float32)

    def _diversity_stats(self) -> np.ndarray:
        """
        Distribições atuais de cada dimensão de diversidade no sys_hist.
        Informa diretamente ao agente o que está sobre/sub-representado.
        """
        return np.concatenate([
            self._get_dist('genres',       N_GENRES,  'genre'),
            self._get_dist('gender_group', 3,         'gender'),
            self._get_dist('race_group',   3,         'race'),
            self._get_dist('region',       N_REGION,  'region'),
            self._get_dist('langs',        N_LANGS,   'lang'),
        ])

    def _obs(self) -> np.ndarray:
        return np.concatenate([
            self._weighted_mean_feat(self._user_hist, most_recent_first=False),
            self._weighted_mean_feat(self._sys_hist,  most_recent_first=True),
            self._diversity_stats(),
            self._genre_avg_rating,
        ]).astype(np.float32)

    def _compute_genre_avg_rating(self) -> np.ndarray:
        """Rating médio por gênero do usuário atual, normalizado 0-1. Precomputado no reset."""
        user_rats = self.ratings.get(self._user_id, {})
        g_sum = np.zeros(N_GENRES, dtype=np.float32)
        g_cnt = np.zeros(N_GENRES, dtype=np.float32)
        for mid, rat in user_rats.items():
            if mid in self.features:
                g = self.features[mid]['genres']
                g_sum += g * rat
                g_cnt += g
        avg = np.where(g_cnt > 0, g_sum / g_cnt, 2.5)
        return (avg / 5.0).astype(np.float32)

    # ── Reset ─────────────────────────────────────────────────────────────────

    def reset(self, *, seed=None, options=None):
        super().reset(seed=seed)
        self._user_id = random.choice(self.user_ids)
        user_rats     = self.ratings.get(self._user_id, {})

        rated           = list(user_rats.keys())
        self._user_hist = deque(rated[-self.HIST_USER:], maxlen=self.HIST_USER)

        prev           = self.rec_history.get(self._user_id, [])
        self._sys_hist = deque(prev[-self.HIST_SYS:], maxlen=self.HIST_SYS)

        self._seen             = set(user_rats.keys())
        self._step_n           = 0
        self._last_targets     = {}
        self._genre_avg_rating = self._compute_genre_avg_rating()

        return self._obs(), {}

    # ── Seleção de filme (dois estágios) ──────────────────────────────────────

    def _select_movie(self, genre_p, gender_p, race_p, region_p, lang_p):
        """
        Estágio 1: cada agente de diversidade amostra uma categoria alvo.
        Estágio 2: candidatos são pontuados com boost 3× para filmes que
                   correspondem às categorias alvo.
        """
        tgt_gender = int(np.random.choice(3,        p=gender_p))
        tgt_race   = int(np.random.choice(3,        p=race_p))
        tgt_region = int(np.random.choice(N_REGION, p=region_p))
        self._last_targets = {'gender': tgt_gender, 'race': tgt_race, 'region': tgt_region}

        candidates = [m for m in all_movie_ids if m not in self._seen]
        if not candidates:
            return None
        if len(candidates) > self.MAX_CAND:
            candidates = random.sample(candidates, self.MAX_CAND)

        scores = np.zeros(len(candidates))
        for i, mid in enumerate(candidates):
            f = self.features[mid]

            s_genre = float(np.dot(genre_p, f['genres']))
            s_lang  = float(np.dot(lang_p,  f['langs']))

            gg = f['gender_group']
            s_gender = float(gg[tgt_gender]) / (gg.sum() + 1e-8)
            rg = f['race_group']
            s_race   = float(rg[tgt_race])   / (rg.sum() + 1e-8)
            reg = f['region']
            s_region = float(reg[tgt_region]) / (reg.sum() + 1e-8)

            scores[i] = s_genre + s_lang + 3.0 * (s_gender + s_race + s_region)

        sel_p = _softmax(scores * 2.0)
        return candidates[int(np.random.choice(len(candidates), p=sel_p))]

    # ── Step ──────────────────────────────────────────────────────────────────

    def step(self, actions: dict):
        genre_p  = _softmax(np.array(actions['genre'],    dtype=np.float32))
        gender_p = _softmax(np.array(actions['gender'],   dtype=np.float32))
        race_p   = _softmax(np.array(actions['race'],     dtype=np.float32))
        region_p = _softmax(np.array(actions['region'],   dtype=np.float32))
        lang_p   = _softmax(np.array(actions['language'], dtype=np.float32))

        mid = self._select_movie(genre_p, gender_p, race_p, region_p, lang_p)
        self._step_n += 1

        if mid is None:
            rewards = {k: 0.0 for k in ['genre', 'gender', 'race', 'region', 'language']}
            return self._obs(), rewards, True, False, {}

        tgt = self._last_targets
        rewards = {
            'genre':    self._r_genre(mid),
            'gender':   self._r_diversity(mid, gender_p, 'gender_group', 3,        tgt['gender'], 'gender'),
            'race':     self._r_diversity(mid, race_p,   'race_group',   3,        tgt['race'],   'race'),
            'region':   self._r_diversity(mid, region_p, 'region',       N_REGION, tgt['region'], 'region'),
            'language': self._r_language(mid, lang_p),
        }

        self._sys_hist.appendleft(mid)
        self._seen.add(mid)

        done = self._step_n >= self.MAX_STEPS
        return self._obs(), rewards, done, False, {'movie_id': mid}

    # ── Helpers ───────────────────────────────────────────────────────────────

    def _movie_dist(self, movie_id: int, feat_key: str, n: int):
        """Normaliza o vetor de atributos do filme para distribuição de probabilidade."""
        f   = self.features.get(movie_id, {})
        vec = f.get(feat_key, np.zeros(n)).astype(np.float32)
        s   = vec.sum()
        return vec / s if s > 0 else None

    # ── Recompensas ───────────────────────────────────────────────────────────

    def _r_genre(self, movie_id: int) -> float:
        """
        R1 = 0.6 × preferência_do_usuário + 0.4 × raridade_do_gênero_na_sessão
        Calculada sobre o FILME SELECIONADO.
        """
        movie_dist = self._movie_dist(movie_id, 'genres', N_GENRES)
        if movie_dist is None:
            return 0.3

        pref        = float(np.dot(movie_dist, self._genre_avg_rating))
        hist        = self._get_dist('genres', N_GENRES, 'genre')
        session_div = max(1.0 - float(np.dot(movie_dist, hist)), 0.0)

        return 0.6 * pref + 0.4 * session_div

    def _r_diversity(self, movie_id: int, probs: np.ndarray,
                     feat_key: str, n: int, target_idx: int, prior_key: str) -> float:
        """
        R2, R3, R4 — Recompensa de diversidade baseada no FILME SELECIONADO.
        (a) sub-representação: grupo do filme é raro no histórico?
        (b) alinhamento: filme pertence ao grupo que o agente escolheu?
        """
        movie_dist = self._movie_dist(movie_id, feat_key, n)
        if movie_dist is None:
            return 0.3

        hist     = self._get_dist(feat_key, n, prior_key)
        underrep = max(1.0 - float(np.dot(movie_dist, hist)), 0.0)
        align    = float(movie_dist[target_idx]) if target_idx < n else 0.0

        return 0.6 * underrep + 0.4 * align

    def _r_language(self, movie_id: int, lang_p: np.ndarray) -> float:
        """
        R5 — Diversidade de idioma baseada no FILME SELECIONADO.
        (a) sub-representação do idioma no histórico
        (b) alinhamento com lang_p do agente
        """
        movie_dist = self._movie_dist(movie_id, 'langs', N_LANGS)
        if movie_dist is None:
            return 0.3

        hist      = self._get_dist('langs', N_LANGS, 'lang')
        underrep  = max(1.0 - float(np.dot(movie_dist, hist)), 0.0)
        alignment = float(np.dot(lang_p, movie_dist))

        return 0.6 * underrep + 0.4 * alignment


# Teste rápido do ambiente
env_train = MovieRecommendationEnv(movie_features, ratings, rec_history, train_users)
obs, _    = env_train.reset()
print(f'Obs shape: {obs.shape}  (esperado: ({OBS_DIM},))')

dummy_actions = {
    'genre':    np.ones(N_GENRES),
    'gender':   np.ones(3),
    'race':     np.ones(3),
    'region':   np.ones(N_REGION),
    'language': np.ones(N_LANGS),
}
obs2, rewards, done, _, info = env_train.step(dummy_actions)
print(f'Recompensas do passo de teste: {rewards}')
print(f'Filme recomendado: {info.get("movie_id")}')
````

## 8. Agentes (Actor-Critic)

Cada agente usa uma rede **Actor-Critic**:

```
obs (OBS_DIM) → [Linear 256 → ReLU] → [Linear 128 → ReLU]
                                           │                   │
                              policy_head (action_dim)   value_head (1)
                              (logits → softmax)         estimativa V(s)
```

Os **cinco agentes** rodam em paralelo sobre a mesma observação e são atualizados com suas respectivas recompensas:

| Agente | Output (logits) | Ações |
|--------|-----------------|-------|
| 1 — Gênero | `n_genres` | Top-3 gêneros com pesos [50%, 30%, 20%] |
| 2 — Dir. Gênero | 3 | Probabilidade sobre [MALE, FEMALE, MIX] |
| 3 — Dir. Raça | 3 | Probabilidade sobre [WHITE, NO_WHITE, MIX] |
| 4 — Dir. Região | `n_regions` | Probabilidade sobre 9 regiões |
| 5 — Idioma | `n_langs` | Top-3 idiomas com pesos [50%, 30%, 20%] |

````python
class AgentNet(nn.Module):
    """
    Rede Actor-Critic para um agente de recomendação.

    - Actor (policy_head): gera logits sobre as ações possíveis.
    - Critic (value_head): estima o valor do estado atual V(s).

    O treinamento usa Actor-Critic (A2C) com bônus de entropia para encorajar
    exploração no início do treino.
    """
    def __init__(self, obs_dim: int, action_dim: int, name: str = ''):
        super().__init__()
        self.name = name

        self.shared = nn.Sequential(
            nn.Linear(obs_dim, 256), nn.ReLU(),
            nn.Linear(256, 128),    nn.ReLU(),
        )
        self.policy_head = nn.Linear(128, action_dim)
        self.value_head  = nn.Linear(128, 1)

    def forward(self, obs: torch.Tensor):
        h      = self.shared(obs)
        logits = self.policy_head(h)
        value  = self.value_head(h).squeeze(-1)
        return logits, value

    def act(self, obs: torch.Tensor, temperature: float = 1.0):
        """Amostra uma ação e retorna (probs, idx, log_prob, value)."""
        logits, value = self.forward(obs)
        probs         = torch.softmax(logits / temperature, dim=-1)
        dist          = torch.distributions.Categorical(probs)
        idx           = dist.sample()
        return probs, idx, dist.log_prob(idx), value


# Cria os cinco agentes
agents = {
    'genre':    AgentNet(OBS_DIM, N_GENRES,  name='Agente 1 — Gênero'),
    'gender':   AgentNet(OBS_DIM, 3,          name='Agente 2 — Gênero Dir.'),
    'race':     AgentNet(OBS_DIM, 3,          name='Agente 3 — Raça Dir.'),
    'region':   AgentNet(OBS_DIM, N_REGION,   name='Agente 4 — Região Dir.'),
    'language': AgentNet(OBS_DIM, N_LANGS,    name='Agente 5 — Idioma'),
}

optimizers = {
    name: torch.optim.Adam(agent.parameters(), lr=3e-4)
    for name, agent in agents.items()
}

total_params = sum(p.numel() for a in agents.values() for p in a.parameters())
for name, agent in agents.items():
    n = sum(p.numel() for p in agent.parameters())
    print(f'{agent.name:<30} — {n:,} parâmetros')
print(f'{"TOTAL":<30} — {total_params:,} parâmetros')
````

## 9. Treinamento (Actor-Critic A2C)

### Algoritmo por agente

```
Para cada episódio:
  obs, _ = env.reset()
  Para cada passo (até MAX_STEPS):
    probs, idx, log_prob, V(s) = agente.act(obs, temperature)
    actions[agente] = probs
    obs', rewards, done = env.step(actions)
    Para cada agente:
      advantage  = R_normalizado - V(s)
      L_policy   = -log_prob × advantage
      L_valor    = MSE(V(s), R_normalizado)
      L_entropia = -H(probs)
      loss       = L_policy + 0.5 × L_valor - 0.05 × L_entropia
```

### Melhorias v2

| Parâmetro | v1 | v2 | Motivo |
|-----------|----|----|--------|
| `entropy_coef` | 0.01 | **0.05** | Previne convergência prematura para o grupo dominante |
| temperatura | 1.5 → 1.0 | **2.0 → 1.0** | Exploração mais ampla no início |
| normalização | nenhuma | **Welford online** | Escalas comparáveis entre os 5 agentes |
| `n_episodes` | 3000 | **5000** | Mais tempo para aprender padrões raros |

> ⚠️ Use `N_EPISODES = 500` para um teste rápido (∼2 min). Para resultados com diversidade, use ≥ 5000.</cell>
<parameter name="cell_type">markdown

````python
def train(
    env: MovieRecommendationEnv,
    agents: dict,
    optimizers: dict,
    n_episodes: int = 5000,
    gamma: float = 0.99,
    entropy_coef: float = 0.05,
    log_every: int = 200,
) -> dict:
    """
    Treina os 5 agentes em paralelo com Actor-Critic (A2C).

    v2: entropy_coef=0.05, temperatura 2.0→1.0, normalização online por Welford.
    """
    history = {name: [] for name in agents}

    # Normalização de recompensa online por agente (Welford)
    rew_mean  = {n: 0.0 for n in agents}
    rew_M2    = {n: 0.0 for n in agents}
    rew_count = {n: 0   for n in agents}

    def normalize_reward(r: float, name: str) -> float:
        n = rew_count[name] + 1
        rew_count[name]  = n
        delta            = r - rew_mean[name]
        rew_mean[name]  += delta / n
        delta2           = r - rew_mean[name]
        rew_M2[name]    += delta * delta2
        std = math.sqrt(rew_M2[name] / max(n - 1, 1) + 1e-8)
        return (r - rew_mean[name]) / std

    for episode in range(1, n_episodes + 1):
        obs, _ = env.reset()
        obs_t  = torch.tensor(obs, dtype=torch.float32)

        ep_rewards = {name: [] for name in agents}

        while True:
            agent_data = {}
            actions    = {}

            for name, agent in agents.items():
                # Temperatura decai de 2.0 → 1.0 ao longo do treino
                temp = max(1.0, 2.0 - episode / n_episodes)
                probs, idx, log_prob, value = agent.act(obs_t, temperature=temp)
                agent_data[name] = (probs, log_prob, value)
                actions[name]    = probs.detach().numpy()

            obs_next, rewards, done, _, info = env.step(actions)
            obs_next_t = torch.tensor(obs_next, dtype=torch.float32)

            for name, agent in agents.items():
                probs, log_prob, value = agent_data[name]
                raw_reward = float(rewards[name])
                ep_rewards[name].append(raw_reward)

                # Normaliza por agente para gradientes comparáveis entre os 5
                reward = normalize_reward(raw_reward, name)

                if not done:
                    with torch.no_grad():
                        _, v_next = agent(obs_next_t)
                    target = reward + gamma * v_next.item()
                else:
                    target = reward

                advantage   = target - value.item()
                policy_loss = -log_prob * advantage
                value_loss  = F.mse_loss(value, torch.tensor(target))
                entropy     = torch.distributions.Categorical(probs).entropy()

                loss = policy_loss + 0.5 * value_loss - entropy_coef * entropy

                optimizers[name].zero_grad()
                loss.backward()
                nn.utils.clip_grad_norm_(agent.parameters(), max_norm=1.0)
                optimizers[name].step()

            obs_t = obs_next_t
            if done:
                break

        for name in agents:
            history[name].append(np.mean(ep_rewards[name]))

        if episode % log_every == 0:
            means = {n: np.mean(h[-log_every:]) for n, h in history.items()}
            print(f'Ep {episode:>4} │ ' +
                  '  '.join(f'{n}={v:.3f}' for n, v in means.items()))

    return history


N_EPISODES = 5000   # ← altere para 500 para teste rápido

print(f'Treinando {len(agents)} agentes por {N_EPISODES} episódios...')
history = train(env_train, agents, optimizers, n_episodes=N_EPISODES)
````

````python
def evaluate_diversity(agents: dict, n_users: int = 30, n_recs: int = 10) -> dict:
    """
    Avalia a diversidade real das recomendações em usuários de teste.
    Compara a distribuição das recomendações com os priors do catálogo.
    """
    gender_counts  = np.zeros(3)
    race_counts    = np.zeros(3)
    region_counts  = np.zeros(N_REGION)
    genre_counts   = np.zeros(N_GENRES)
    lang_counts    = np.zeros(N_LANGS)
    total          = 0

    sample_users = random.sample(test_users, min(n_users, len(test_users)))
    for user_id in sample_users:
        recs = recommend(user_id, agents, n=n_recs)
        for r in recs:
            mid = r['movie_id']
            f   = movie_features.get(mid, {})

            gg = f.get('gender_group', np.zeros(3))
            if gg.sum() > 0:
                gender_counts[np.argmax(gg)] += 1

            rg = f.get('race_group', np.zeros(3))
            if rg.sum() > 0:
                race_counts[np.argmax(rg)] += 1

            reg = f.get('region', np.zeros(N_REGION))
            if reg.sum() > 0:
                region_counts[np.argmax(reg)] += 1

            genre_counts += f.get('genres', np.zeros(N_GENRES))
            lang_counts  += f.get('langs',  np.zeros(N_LANGS))
            total += 1

    print(f'\n{"═"*65}')
    print(f'  AVALIAÇÃO DE DIVERSIDADE  ({total} filmes, {n_users} usuários)')
    print(f'{"═"*65}')

    def _pct(arr):
        s = arr.sum()
        return arr / s * 100 if s > 0 else arr

    print('\nGênero do Diretor:')
    pct = _pct(gender_counts)
    for i, g in enumerate(GENDER_GROUPS):
        bar = '█' * int(pct[i] / 2)
        base = DATASET_PRIORS['gender'][i] * 100
        flag = ' ← ↑ diversidade!' if pct[i] > base else ''
        print(f'  {g:<12}: {pct[i]:5.1f}%  {bar:<30}  (catálogo: {base:.1f}%){flag}')

    print('\nRaça do Diretor:')
    pct = _pct(race_counts)
    for i, g in enumerate(RACE_GROUPS):
        bar = '█' * int(pct[i] / 2)
        base = DATASET_PRIORS['race'][i] * 100
        flag = ' ← ↑ diversidade!' if g != 'WHITE' and pct[i] > base else ''
        print(f'  {g:<12}: {pct[i]:5.1f}%  {bar:<30}  (catálogo: {base:.1f}%){flag}')

    print('\nTop-5 Gêneros de Filmes:')
    top5 = np.argsort(genre_counts)[-5:][::-1]
    for i in top5:
        pct_g = genre_counts[i] / total * 100 if total > 0 else 0
        print(f'  {genre_cols[i].replace("genre_",""):<20}: {pct_g:.1f}%')

    print('\nTop-5 Idiomas:')
    top5l = np.argsort(lang_counts)[-5:][::-1]
    for i in top5l:
        pct_l = lang_counts[i] / total * 100 if total > 0 else 0
        base_l = DATASET_PRIORS['lang'][i] * 100
        flag = ' ← ↑ diversidade!' if pct_l < base_l * 0.8 else ''
        print(f'  {lang_cols[i].replace("lang_",""):<20}: {pct_l:.1f}%  (catálogo: {base_l:.1f}%){flag}')

    return {
        'gender': gender_counts,
        'race': race_counts,
        'region': region_counts,
        'genre': genre_counts,
        'lang': lang_counts,
        'total': total,
    }


print('Avaliando diversidade nas recomendações...')
div_stats = evaluate_diversity(agents, n_users=30, n_recs=10)
````

## 10. Simulação de Recomendação

Com os agentes treinados, geramos recomendações para um usuário de teste.

Para cada recomendação, o sistema exibe:
- **Top-3 gêneros** (Agente 1) com probabilidades aproximadas [50%, 30%, 20%]
- **Distribuição de gênero do diretor** (Agente 2): [MALE%, FEMALE%, MIX%]
- **Distribuição de raça do diretor** (Agente 3): [WHITE%, NO_WHITE%, MIX%]
- **Top-3 regiões do diretor** (Agente 4)
- **Top-3 idiomas** (Agente 5)

````python
def recommend(user_id: int, agents: dict, n: int = 10) -> list:
    """
    Gera n recomendações para um usuário usando os 5 agentes treinados.
    """
    for agent in agents.values():
        agent.eval()

    # Ambiente de inferência para o usuário de teste
    env_infer = MovieRecommendationEnv(
        movie_features, ratings, rec_history, [user_id]
    )
    env_infer.MAX_STEPS = n

    obs, _ = env_infer.reset()
    recs   = []

    with torch.no_grad():
        for step in range(n):
            obs_t   = torch.tensor(obs, dtype=torch.float32)
            actions = {}
            probs_display = {}

            for name, agent in agents.items():
                logits, _ = agent(obs_t)
                p = torch.softmax(logits, dim=-1).numpy()
                actions[name]       = p
                probs_display[name] = p

            obs, rewards, done, _, info = env_infer.step(actions)

            if 'movie_id' not in info:
                continue

            mid = info['movie_id']
            row = enc_df[enc_df['movieid'] == mid]
            title = row['title'].values[0] if len(row) > 0 else f'Filme {mid}'

            # Top-3 gêneros
            gp   = probs_display['genre']
            top3g = [(genre_cols[i].replace('genre_',''), f'{gp[i]:.1%}')
                     for i in np.argsort(gp)[-3:][::-1]]

            # Top-3 idiomas
            lp   = probs_display['language']
            top3l = [(lang_cols[i].replace('lang_',''), f'{lp[i]:.1%}')
                     for i in np.argsort(lp)[-3:][::-1]]

            # Top-3 regiões
            rp   = probs_display['region']
            top3r = [(REGION_NAMES[i], f'{rp[i]:.1%}')
                     for i in np.argsort(rp)[-3:][::-1]]

            recs.append({
                'step':        step + 1,
                'movie_id':    mid,
                'title':       title,
                'top3_genres': top3g,
                'top3_langs':  top3l,
                'top3_regions':top3r,
                'gender_dist': dict(zip(GENDER_GROUPS, probs_display['gender'].tolist())),
                'race_dist':   dict(zip(RACE_GROUPS,   probs_display['race'].tolist())),
                'rewards':     rewards,
            })

            if done:
                break

    return recs


# Seleciona um usuário de teste e gera recomendações
TEST_USER = test_users[0]
print(f'Gerando recomendações para usuário {TEST_USER}...')
print(f'  Filmes avaliados: {len(ratings.get(TEST_USER, {})):,}')

recs = recommend(TEST_USER, agents, n=10)
````
